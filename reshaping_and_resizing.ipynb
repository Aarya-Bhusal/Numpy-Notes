# Reshaping and Resizing Arraays

```python
import numpy as np
```

```python
arr = np.arange(1,26)
arr
```

```python
# Original Array and Reshaped Array must have equal number of elements
arr.reshape((5,5))
```

```python
np.resize(arr, (6,5))
# Unlike reshape resized array can have more elements than the original one but the elements repeats
```
