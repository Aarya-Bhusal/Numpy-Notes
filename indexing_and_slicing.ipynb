# Indexing and Slicing of Vectors

```python
import numpy as np
```

```python
arr = np.arange(1,11)
arr
```

```python
# Indexing is same as like in python
arr[3]
```

```python
arr[3:8] # The first number is inclusive and the second one is exclusive
```

```python
# For last five elements
arr[5:]
```

```python
# For first five elements
arr[:5]
```

```python
# To print the array with gaps(steps)
arr[::2]
```

# Indexing and Slicing of Matrix

```python
arr = np.arange(1,26).reshape((5,5))
arr
```

```python
arr[0] # If we will provide a single index for a matrix we will get a vector at that index
```

```python
arr[0,1] # To get a single element we must give a proper double index in a matrix
```

```python
# To get a specific area in a matrix
arr[1:4, 1:4]
# [from_row(inclusive):to_row(exclusive), from_col(inclusive):to_col(exclusive)]
```

```python
arr[2:, 3:]
```

```python
# To get a single column in a matrix
arr[:, 1]
```

# Boolean Indexing

```python
bool_index = arr % 2 == 0 
bool_index
```

```python
# CHange all even number to 0
arr[bool_index] = 0
arr
```
