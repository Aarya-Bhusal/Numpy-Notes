# Exercises of Numpy

```python
import numpy as np
```

```python
# Valid sudoko
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
sudoko = True
row, col = s.shape

if(row==9 and col==9):
    # Check if every row has unique numbers
    for i in range(9):
        r = s[i]
        if(len(np.unique(r))!=9):
            sudoko = False
            break

    # Check if every column has unique numbers
    for i in range(9):
        c = s[:,i]
        if(len(np.unique(c))!=9):
            sudoko = False
            break
    # Check if every 3x3 box has unique numbers
    for row in range(0,9,3):
        for col in range(0,9,3):
            box = s[row:row+3, col:col+3]
            if(len(np.unique(box))!=9):
                sudoko = False
                break
else:
    sudoko = False

if(sudoko):
    print("Valid Sudoko")
else:
    print("Invalid Sudoko")
```

# General Questions

```python
# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
data.shape
```

```python
#Find the average age of students.
age_row = data[:, 0:1]
age_row.mean()
```

```python
#Extract Math marks of all students.
maths_marks = data[:, 1:2]
maths_marks
```

```python
#Find the highest Science mark.
science_marks = data[:, 2:]
science_marks.max()
```

```python
#Get details of the student who scored more than 90 in Math.
for row in range(len(data)):
    print(data[row]) if data[row, 1]>90 else None
```

```python
#Increase Math marks of all students by 5.
maths_marks = data[:, 1]
maths_marks += 5
print(data)
```

```python
#Find how many students are younger than 19.
age_col = data[:, 0]
bool_index = age_col<19
print(sum(bool_index))
```

```python
#Calculate the average marks in each subject (column-wise mean).
science_col = data[:, 1]
maths_col = data[:, 2]
print(f"Average Marks in Science : {np.mean(science_col)}")
print(f"Average Marks in Maths : {np.mean(maths_col)}")
```

```python
#Get data of students who scored at least 80 in both subjects.
for row in range(len(data)):
    print(data[row]) if (data[row, 1]>=80 and data[row, 2]>=80) else None
```
