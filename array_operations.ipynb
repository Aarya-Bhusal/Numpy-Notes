**Broadcasting**

```python
arr = np.arange(1,11)
arr
```

```python
# Add 5 to every element in the the array
arr + 5
```

```python
# Subract 5 from every element in the the array
arr - 5
```

**Deep and Shallow Copy**

```python
# Deep Copy
a = np.arange(1,11)
print(f"a = {a}")
b = a
b[0] = 99
print(f"b = {b}")
print(f"a = {a}") # Here changing b has also changed a because they share the same memory location
```

```python
# Shallow Copy
a = np.arange(1,11)
print(f"a = {a}")
b = a.copy()
b[0] = 99
print(f"b = {b}")
print(f"a = {a}") # Here changing b doesn't change a because they b is in seperate memory location
```

**Matrix Operations**

```python
A = np.arange(1,10).reshape(3,3)
A
```

```python
B = np.arange(11,20).reshape(3,3)
B
```

```python
# Dot product of two matrix
A @ B
```

```python
# Transposing a matrix
A.T
```

**Advance Array Manipulation**

```python
# Stacking arrays vertically
a = np.array([1,2,3,4,5])
b = np.array([6,7,8,9,10])
np.vstack((a,b))
```

```python
# Stacking arrays horizontally
a = np.array([1,2,3,4,5])
b = np.array([6,7,8,9,10])
np.hstack((a,b))
```

```python
# Stacking arrays in columns 
a = np.array([1,2,3,4,5])
b = np.array([6,7,8,9,10])
np.column_stack((a,b))
```

**Spliting Array**

```python
arr = np.arange(1,17).reshape(4,4)
arr
```

```python
np.hsplit(arr, 2)
```

```python
np.vsplit(arr,2)
```

*The main array is divided into given parameter equally horizontally or vertically but the number of elements must be divided exactly by parameter. The output is also a array that has all the splitted arrays but the splitted arrays are also 2D.*
